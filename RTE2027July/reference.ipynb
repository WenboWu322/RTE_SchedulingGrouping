```python
# xlsx to csv and make sure the titles are correct
import pandas as pd

df = pd.read_excel("operations_CR_2005_2024.xlsx")

df.columns = (
    df.columns.astype(str)
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
    .str.strip()
)

df = df.replace(r"\n|\r", " ", regex=True)

df.to_csv("operations_CR_2005_2024.csv", index=False, sep=",", encoding="utf-8-sig", decimal=".")
```

```python
# calculate the leakage of each csem and the entire station
import pandas as pd

df = pd.read_csv("fused data.csv", encoding="utf-8-sig")

cr_cols = [f"Nb CR {y}" for y in range(2010, 2025)]
cr_cols = [c for c in cr_cols if c in df.columns]

for c in cr_cols + ["Masse rejetée à chaque CR (kg)"]:
    df[c] = pd.to_numeric(df[c].astype(str).str.replace(",", ".", regex=False), errors="coerce")

df["leakage"] = df[cr_cols].sum(axis=1) * df["Masse rejetée à chaque CR (kg)"]
total_leakage = df["leakage"].sum()

print("Total leakage (kg):", total_leakage)

result = df[["N Equipement", "leakage"]].copy()
print(result)
```
