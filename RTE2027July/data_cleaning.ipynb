```python
import pandas as pd

df = pd.read_csv("operations_CR_2005_2024.csv")

cols = [f"Nb CR {year}" for year in range(2010, 2025)]

# 把空字符串当成缺失值
df[cols] = df[cols].replace("", pd.NA)

# 其余列也保留：同组取第一个非空值
def first_non_null(s):
    s = s.dropna()
    return s.iloc[0] if len(s) > 0 else pd.NA

agg_dict = {col: first_non_null for col in cols}
for col in df.columns:
    if col not in ["N Equipement"] + cols:
        agg_dict[col] = first_non_null

result = df.groupby("N Equipement", as_index=False).agg(agg_dict)

result.to_csv("fused data.csv", index=False, encoding="utf-8-sig")

# 统计不同 N Equipement 的行数
count_by_equipment = df["N Equipement"].value_counts(dropna=False).sort_index()

print(count_by_equipment)
```
