```python
# new P sensitivity plot with all seeds M1
# plot sensitivity P for different seeds
import pandas as pd
import glob
import os

# 指定文件夹路径
folder_path = 'Sensitivity_P_MIMAR_Wiener'  # 替换为你的文件夹路径


# 查找所有匹配的CSV文件
pattern = os.path.join(folder_path, "Sensitivity_P10-200_seed*_MIMAR_Wiener.csv")
csv_files = glob.glob(pattern)

if not csv_files:
    print("未找到匹配的CSV文件，请检查文件名模式。")
else:
    dfs = []
    for file in csv_files:
        df = pd.read_csv(file)
        dfs.append(df)

    # 垂直合并所有DataFrame（假设行数相同，对应行合并）
    combined_df = pd.concat(dfs, ignore_index=False)

    # 按行索引分组，计算每列均值（对应行平均）
    avg_df = combined_df.groupby(combined_df.index).mean()

    # 保存结果为average_results.csv，保留列名
    # avg_df.to_csv('average_results.csv')
    # print("平均结果已保存到 average_results.csv")
    print(avg_df)  # 预览前几行
```
