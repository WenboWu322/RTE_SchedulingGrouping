```python
import pandas as pd
import glob
import os
import numpy as np

folder_path = 'Sensitivity_P_MIMAR'
pattern = os.path.join(folder_path, "Sensitivity_P10-200_seed*_MIMAR.csv")
csv_files = glob.glob(pattern)

if not csv_files:
    print("未找到文件。")
else:
    dfs = [pd.read_csv(file) for file in csv_files]
    
    # 获取所有列（指标）
    all_cols = set(col for df in dfs for col in df.columns)
    all_cols = list(all_cols)
    
    # 初始化三个表
    mean_table = pd.DataFrame(index=dfs[0].index)  # 行索引从第一个文件
    var_table = pd.DataFrame(index=dfs[0].index)
    std_table = pd.DataFrame(index=dfs[0].index)
    
    for col in all_cols:
        col_dfs = [df[col] for df in dfs if col in df.columns]
        if len(col_dfs) == len(dfs):  # 确保所有文件都有该列
            col_combined = pd.concat(col_dfs, axis=1)
            mean_table[col] = col_combined.mean(axis=1)
            var_table[col] = col_combined.var(axis=1)
            std_table[col] = col_combined.std(axis=1)
    
    # 保存三个表
    mean_table.to_csv('mean_table.csv')
    var_table.to_csv('var_table.csv')
    std_table.to_csv('std_table.csv')
    
    print("均值表: mean_table.csv")
    print("方差表: var_table.csv")
    print("标准差表: std_table.csv")
    print("均值表预览:\n", mean_table.head())
```
